```python
# Based on typical CLIR performance, here are recommended hybrid weights:

recommendations = {
    "balanced": {
        "BM25": 0.33,
        "Fuzzy": 0.33,
        "Semantic": 0.34,
        "reason": "Equal weight to all methods - good starting point"
    },
    "semantic_focused": {
        "BM25": 0.2,
        "Fuzzy": 0.2,
        "Semantic": 0.6,
        "reason": "Prioritize semantic embeddings (best for cross-lingual)"
    },
    "keyword_focused": {
        "BM25": 0.5,
        "Fuzzy": 0.3,
        "Semantic": 0.2,
        "reason": "Prioritize exact keyword matching (good for known entities)"
    },
    "balanced_with_fuzzy": {
        "BM25": 0.3,
        "Fuzzy": 0.4,
        "Semantic": 0.3,
        "reason": "Handle typos/transliteration (good for user queries)"
    },
}

print("\n" + "="*60)
print("RECOMMENDED HYBRID WEIGHT STRATEGIES")
print("="*60)

for strategy_name, weights in recommendations.items():
    reason = weights.pop("reason")
    print(f"\n📊 Strategy: {strategy_name.upper()}")
    print(f"   Reason: {reason}")
    print(f"   Weights:")
    for method, weight in weights.items():
        print(f"     - {method}: {weight:.0%}")

print("\n" + "="*60)
print("\nTo choose the best strategy:")
print("1. Review the test results above")
print("2. Pick the strategy that matches your needs")
print("3. We'll implement it in the search system")
```
